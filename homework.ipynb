```python
# Task 1. Generate a text file called zad1.txt containing random negative temperatures.
from random import randint

with open("zad1.txt", "w") as data:
    s = ""
    for _ in range(100):
        s += f"{randint(-100,-1)}\n"
    data.write(s)
```

```python
# Task 2. Find and display the three lowest temperatures.

with open("zad1.txt","r") as data:
    t = []
    for linie in data:
        t.append(int(linie))

temperatures = t.copy()

def bubble_sort(mylist:list)->list:
    n = len(mylist)
    for i in range(n-1):
        for j in range(n-i-1):
            if mylist[j] > mylist[j+1]:
                mylist[j], mylist[j+1] = mylist[j+1], mylist[j]
    return mylist

sorted_temperatures = bubble_sort(temperatures)
print(sorted_temperatures)

lowest = []
i  = 0
while len(lowest) < 3:
    if not sorted_temperatures[i] in lowest:
        lowest.append(sorted_temperatures[i])
    i += 1

print("The three lowest temperatures:")
print(lowest[0])
print(lowest[1])
print(lowest[2])
```

```python
import random


with open('bin.txt', 'w') as f:
    for _ in range(1000):
        val = random.randint(1, 10000)
        bin_str = bin(val)[2:]         
        f.write(bin_str + '\n')

with open('bin.txt', 'r') as bin_f, open('zad2.txt', 'w') as zad2_f:
    for line in bin_f:
        line = line.strip()
        if line:
            num = int(line, 2)
            zad2_f.write(str(num) + '\n')

with open('zad2.txt', 'r') as f:
    numbers = [int(line.strip()) for line in f if line.strip()]

with open('zad6.txt', 'w') as f:
    for num in numbers:
        f.write(f"{num} - {bin(num)[2:]}\n")

with open('zad7.txt', 'w') as f:
    for num in numbers:
        if num % 3 == 0 and num % 5 == 0:
            f.write(f"{num}\n")

with open('zad8.txt', 'w') as f:
    for num in numbers:
        digit_sum = sum(int(digit) for digit in str(num))
        f.write(f"{num} - {digit_sum}\n")


reversed_numbers = []
for i in range(len(numbers) - 1, -1, -1):
    reversed_numbers.append(numbers[i])

with open('zad9.txt', 'w') as f:
    for num in reversed_numbers:
        f.write(f"{num}\n")


with open('zad10.txt', 'w') as f:
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == 100:
                f.write(f"{numbers[i]} - {numbers[j]}\n")
```
